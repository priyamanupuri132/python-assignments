```python
marks = [78, 45, 92, 33, 67, 88, 51, 29]

passed = 0

for mark in marks:
    if mark >= 35:
        passed = passed + 1

print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Class average:", sum(marks) / len(marks))
print("Students who passed:", passed)
```

```python
shopping_list = []
item = ""

while item != "done":
    item = input("Enter item: ").lower()

    if item == "done":
        print("Shopping list completed")
    elif item in shopping_list:
        print("Already added")
    elif item != "":
        shopping_list.append(item)

for i in range(len(shopping_list)):
    print(str(i + 1) + ". " + shopping_list[i])

print("Total items:", len(shopping_list))
```

```python
sentence = input("Enter a sentence: ")

vowels = 0
consonants = 0
spaces = 0

for character in sentence.lower():
    if character in "aeiou":
        vowels = vowels + 1
    elif character in "bcdfghjklmnpqrstvwxyz":
        consonants = consonants + 1
    elif character == " ":
        spaces = spaces + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
name = input("Enter your name: ")

lower_name = name.lower()
reversed_name = ""

for character in lower_name:
    reversed_name = character + reversed_name

if lower_name == reversed_name:
    print("Bonus! Your name is a palindrome.")
else:
    print("Your name is not a palindrome.")
```

```python
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

hottest_temperature = temps[0]
hottest_day = days[0]
fahrenheit_temps = []

print("Heat alert days:")

for i in range(len(temps)):
    if temps[i] > 37:
        print(days[i], "-", temps[i], "°C")

    if temps[i] > hottest_temperature:
        hottest_temperature = temps[i]
        hottest_day = days[i]

    fahrenheit = temps[i] * 9 / 5 + 32
    fahrenheit_temps.append(fahrenheit)

print("Hottest day:", hottest_day)
print("Hottest temperature:", hottest_temperature, "°C")
print("Temperatures in Fahrenheit:", fahrenheit_temps)
```

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

if len(words) == 0:
    print("No words entered")
else:
    longest_word = words[0]
    shortest_word = words[0]
    capital_count = 0

    for word in words:
        if len(word) > len(longest_word):
            longest_word = word

        if len(word) < len(shortest_word):
            shortest_word = word

        if word[0].isupper():
            capital_count = capital_count + 1

    print("Longest word:", longest_word)
    print("Shortest word:", shortest_word)
    print("Words starting with capital letter:", capital_count)
```

```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]

positive_accounts = []
negative_accounts = []

for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)

total_owed = -sum(negative_accounts)

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed: Rs.", total_owed)
```

```python
password = input("Enter a password: ")

has_uppercase = False
has_lowercase = False
has_digit = False

for character in password:
    if character.isupper():
        has_uppercase = True
    elif character.islower():
        has_lowercase = True
    elif character.isdigit():
        has_digit = True

conditions_met = 0

if has_uppercase:
    conditions_met = conditions_met + 1

if has_lowercase:
    conditions_met = conditions_met + 1

if has_digit:
    conditions_met = conditions_met + 1

if len(password) >= 8:
    conditions_met = conditions_met + 1

if conditions_met == 4:
    print("Strong")
elif conditions_met >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
full_name = input("Enter your full name: ")
names = full_name.split()

if len(names) == 0:
    print("No name entered")
else:
    initials = ""
    title_name = ""
    surname_name = ""

    for i in range(len(names)):
        part = names[i]
        title_part = part[0].upper() + part[1:].lower()

        initials = initials + part[0].upper() + "."

        if i == 0:
            title_name = title_part
        else:
            title_name = title_name + " " + title_part

        if i < len(names) - 1:
            surname_name = surname_name + part[0].upper() + ". "
        else:
            surname_name = surname_name + title_part

    print("Initials:", initials)
    print("Name in title case:", title_name)
    print("Name with surname in full:", surname_name)
```
