```python
age = int(input("Enter your age: "))

if age < 5:
    print("Your ticket is free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8

print("Total bill: Rs.", bill)
```

```python
total = 0
days = 0

while total < 1000:
    savings = int(input("Enter today's savings: "))
    total = total + savings
    days = days + 1

print("Goal reached in", days, "days!")
print("Total saved: Rs.", total)
```

```python
correct_pin = 4321
attempts = 0
pin = 0

while attempts < 3 and pin != correct_pin:
    pin = int(input("Enter PIN: "))
    attempts = attempts + 1

    if pin == correct_pin:
        print("Access granted")
    else:
        print("Wrong PIN")

if pin != correct_pin:
    print("Card blocked")
```

```python
n = int(input("Enter N: "))

even_count = 0
odd_count = 0

for number in range(1, n + 1):
    if number % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1

print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
```

```python
secret_number = 27
guess = 0
guesses = 0

while guess != secret_number:
    guess = int(input("Guess the number: "))
    guesses = guesses + 1

    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")
    else:
        print("Correct! You took", guesses, "guesses.")
```

```python
bill_number = int(input("Enter bill number: "))

number = bill_number
digit_sum = 0

while number > 0:
    digit = number % 10
    digit_sum = digit_sum + digit
    number = number // 10

print("Sum of digits:", digit_sum)

if digit_sum > 20:
    print("Congratulations! You win a free gift.")
else:
    print("You do not get a free gift.")
```
