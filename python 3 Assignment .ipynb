```python
contacts = {}
choice = ""

while choice != "4":
    print("\n1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ")

        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts saved")
        else:
            for name, phone in contacts.items():
                print(name, ":", phone)

    elif choice == "4":
        print("Contact book closed")

    else:
        print("Invalid choice")
```

```python
marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}
topper = ""
highest_marks = -1

for name, mark in marks.items():
    if mark >= 40:
        print(name, ": Pass")
        passed_students[name] = mark
    else:
        print(name, ": Fail")

    if mark > highest_marks:
        highest_marks = mark
        topper = name

print("Topper:", topper)
print("Topper marks:", highest_marks)
print("Passed students:", passed_students)
```

```python
priya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual_friends = priya_friends.intersection(arjun_friends)
only_priya = priya_friends.difference(arjun_friends)
all_friends = priya_friends.union(arjun_friends)
suggestions = arjun_friends.difference(priya_friends)

print("Mutual friends:", sorted(mutual_friends))
print("Friends only priya has:", sorted(only_priya))
print("All friends:", sorted(all_friends))
print("Friend suggestions for priya:", sorted(suggestions))
```

```python
text = input("Enter text: ")

character_count = {}

for character in text:
    if character != " ":
        character_count[character] = character_count.get(character, 0) + 1

print(character_count)

if len(character_count) == 0:
    print("No characters to count")
else:
    most_frequent = ""
    highest_count = 0

    for character, count in character_count.items():
        if count > highest_count:
            highest_count = count
            most_frequent = character

    print("Most frequent character:", most_frequent,
          "(" + str(highest_count) + " times)")
```

```python
menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0
item = ""

while item != "bill":
    item = input("Enter item or type bill: ").lower()

    if item == "bill":
        print("Order completed")
    elif item in menu:
        total = total + menu[item]
        print("Item added")
    else:
        print("Item not available")

print("Bill before discount: Rs.", total)

if total > 200:
    discount = total * 10 / 100
    total = total - discount
    print("Discount: Rs.", discount)

print("Total bill: Rs.", total)
```

```python
cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

northern_city = cities[0][0]
highest_latitude = cities[0][1]

for city, latitude, longitude in cities:
    print(city, "- Latitude:", latitude, "Longitude:", longitude)

    if latitude > highest_latitude:
        highest_latitude = latitude
        northern_city = city

print("Farthest north city:", northern_city)
print("Highest latitude:", highest_latitude)
```

```python
books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter book name to borrow: ")

if book_name not in books:
    print("Book not found")
elif books[book_name] == 0:
    print("Currently unavailable")
else:
    books[book_name] = books[book_name] - 1
    print("Book issued successfully")

print("Updated books:", books)
```

```python
employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, department in employees:
    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print("Employees by department:", departments)

for department, names in departments.items():
    print(department, "has", len(names), "employees")
```

```python
answer_key = ("B", "C", "A", "D", "B")

score = 0
wrong_questions = set()

for i in range(len(answer_key)):
    answer = input("Enter answer for question " + str(i + 1) + ": ").upper()

    if answer == answer_key[i]:
        print("Question", i + 1, ": Correct")
        score = score + 1
    else:
        print("Question", i + 1, ": Wrong")
        wrong_questions.add(i + 1)

print("Wrong question numbers:", wrong_questions)
print("Final score:", score, "out of 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
